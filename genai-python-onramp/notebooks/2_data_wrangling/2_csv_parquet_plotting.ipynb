# Module 2.2: CSV, Parquet, and Basic Plotting

**Learning Objective:** Learn how to read and write data in common formats (CSV and Parquet) and create basic plots with Pandas, Matplotlib, and Seaborn.

This notebook covers the essentials of data I/O and visualization.

### Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Reading and Writing CSV Files

CSV (Comma-Separated Values) is a very common format for tabular data.

```python
# Create a sample DataFrame
data = {'col1': [1, 2], 'col2': [3, 4]}
df = pd.DataFrame(data)

# Write to a CSV file
df.to_csv('sample.csv', index=False)

# Read from a CSV file
df_from_csv = pd.read_csv('sample.csv')
print(df_from_csv)
```

## Reading and Writing Parquet Files

Parquet is a columnar storage format that is highly efficient for large datasets. You'll need `pyarrow` installed (`pip install pyarrow`).

```python
# Write to a Parquet file
df.to_parquet('sample.parquet')

# Read from a Parquet file
df_from_parquet = pd.read_parquet('sample.parquet')
print(df_from_parquet)
```

## Basic Plotting

Pandas has built-in plotting capabilities that use Matplotlib under the hood. Seaborn is another popular library for creating attractive statistical plots.

```python
# Create a more interesting DataFrame
plot_data = {
    'x': range(10),
    'y': [i**2 for i in range(10)]
}
plot_df = pd.DataFrame(plot_data)

# Simple plot with Pandas
plot_df.plot(x='x', y='y', title='Pandas Plot')
plt.show()
```

```python
# Scatter plot with Seaborn
sns.scatterplot(data=plot_df, x='x', y='y').set_title('Seaborn Scatter Plot')
plt.show()
```

## Recap

- Use `to_csv()` and `read_csv()` for working with CSV files.
- Use `to_parquet()` and `read_parquet()` for efficient storage with Parquet files.
- Pandas, Matplotlib, and Seaborn are powerful tools for data visualization.
