## The Challenge

Your task is to:
1.  **Create a messy CSV file** to simulate a real-world scenario.
2.  **Load the CSV** into a Pandas DataFrame.
3.  **Inspect the data** for issues (missing values, incorrect data types).
4.  **Clean the data.**
5.  **Export the cleaned data** to a new format (e.g., JSON or Parquet).

### Step 1: Create a messy CSV

```python
import pandas as pd
import numpy as np

messy_data = {
    'ID': ['1', '2', '3', '4', '5'],
    'Name': ['Alice', 'Bob', 'Charlie', np.nan, 'Eve'],
    'Age': ['25', '30', '35', '40', 'NaN'],
    'Salary': ['50000', '60000', '70000', '80000', '90000.5']
}
messy_df = pd.DataFrame(messy_data)
messy_df.to_csv('messy_data.csv', index=False)
```

### Step 2: Load and Inspect the Data

```python
df = pd.read_csv('messy_data.csv')
print("Original DataFrame:")
print(df)
print("\nData Types:")
print(df.info())
```

### Step 3: Clean the Data

We need to:
- Handle the missing `Name`.
- Convert `Age` to a numeric type, handling the `NaN` string.
- Convert `Salary` to a numeric type.

```python
# Fill missing name with a placeholder
df['Name'].fillna('Unknown', inplace=True)

# Convert Age to numeric, coercing errors to NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Convert Salary to numeric
df['Salary'] = pd.to_numeric(df['Salary'])

# You might also want to fill the NaN age with a value, like the mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

print("Cleaned DataFrame:")
print(df)
print("\nNew Data Types:")
print(df.info())
```

### Step 4: Export the Cleaned Data

```python
# Export to JSON
df.to_json('cleaned_data.json', orient='records', indent=4)

# You can also export to Parquet
# df.to_parquet('cleaned_data.parquet')
```
