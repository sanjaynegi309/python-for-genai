## Introduction to `pytest`

`pytest` is a popular, powerful, and easy-to-use testing framework for Python.

### Writing a Simple Test

Tests in `pytest` are just functions that start with `test_`. You can use the standard `assert` statement to check for expected outcomes.

````python
# In a file named `test_example.py`:

def add(a, b):
    return a + b

def test_add():
    assert add(2, 3) == 5
    assert add('a', 'b') == 'ab'
````

### Running `pytest`

To run your tests, you simply run the `pytest` command in your terminal. `pytest` will automatically discover and run your test files.

````python
# In your terminal:
# pytest
````

### Testing for Exceptions

You can test that a function raises an exception using `pytest.raises`.

````python
import pytest

def test_zero_division():
    with pytest.raises(ZeroDivisionError):
        1 / 0
````
