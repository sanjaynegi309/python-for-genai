## VS Code for Python Developers

Visual Studio Code is a popular, free, and open-source code editor with excellent Python support.

### Recommended Extensions

- **Python (ms-python.python):** The official Python extension from Microsoft. Provides linting, debugging, code completion, and more.
- **Pylance (ms-python.vscode-pylance):** A high-performance language server for Python that provides features like type checking and auto-imports.
- **Black Formatter (ms-python.black-formatter):** Automatically formats your Python code using the Black code formatter.
- **Ruff (charliermarsh.ruff):** Integrates the Ruff linter for fast and powerful code analysis.
- **Jupyter (ms-toolsai.jupyter):** Provides Jupyter notebook support within VS Code.

## Repository Hygiene: `.gitignore`

A `.gitignore` file specifies intentionally untracked files that Git should ignore. This is crucial for keeping your repository clean and avoiding committing unnecessary files.

```python
# A good .gitignore for a Python project might look like this:
"""
# Virtual Environment
venv/
.venv/

# Python cache
__pycache__/
*.pyc

# IDE / Editor specific
.vscode/
.idea/

# Build artifacts
dist/
build/

# Secrets
.env
"""
```

## Standard Project Structure

A well-organized project structure makes your code easier to understand and maintain. Here's a common structure for a Python application:

```python
"""
my_project/
├── .gitignore
├── pyproject.toml
├── README.md
├── src/
│   └── my_project/
│       ├── __init__.py
│       └── main.py
└── tests/
    └── test_main.py
"""
```
