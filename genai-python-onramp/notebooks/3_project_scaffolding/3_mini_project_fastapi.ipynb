# Module 3.3: Mini-Project - Build a Starter FastAPI Service

**Learning Objective:** Build a simple, production-ready FastAPI service with a basic endpoint and test coverage.

This mini-project will guide you through building the `starter_fastapi` project included in this repository. It brings together concepts like Pydantic, FastAPI, and Pytest.

## The Goal

We will create a simple API with one endpoint: `/ping`. When called, it will return a JSON response with a message.

### Step 1: Project Structure

Navigate to the `starter_fastapi` directory. It has the following structure:
```
starter_fastapi/
├── main.py
├── requirements.txt
└── tests/
    └── test_main.py
```

### Step 2: `main.py` - The FastAPI Application

This file contains the main FastAPI application. We will define the app and the `/ping` endpoint.

````python
# In starter_fastapi/main.py:

from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class PingResponse(BaseModel):
    message: str

@app.get("/ping", response_model=PingResponse)
def ping():
    return {"message": "pong"}
````

### Step 3: `requirements.txt`

This file lists the dependencies needed for the FastAPI service.

````python
# In starter_fastapi/requirements.txt:

fastapi
uvicorn
pytest
httpx
````

### Step 4: `tests/test_main.py` - The Tests

We will write a simple test for our `/ping` endpoint using `pytest` and `httpx`'s `TestClient`.

````python
# In starter_fastapi/tests/test_main.py:

from fastapi.testclient import TestClient
from ..main import app # Note the relative import

client = TestClient(app)

def test_ping():
    response = client.get("/ping")
    assert response.status_code == 200
    assert response.json() == {"message": "pong"}
````

### Step 5: Running the Service and Tests

To run the service:
```bash
cd starter_fastapi
uvicorn main:app --reload
```

To run the tests:
```bash
cd starter_fastapi
pytest
```
