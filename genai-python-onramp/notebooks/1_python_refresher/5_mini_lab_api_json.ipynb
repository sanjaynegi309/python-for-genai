### Step 1: Import necessary libraries

```python
import httpx
import json
import logging
```

### Step 2: Configure logging

```python
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
```

### Step 3: Make the API call

```python
API_URL = "https://jsonplaceholder.typicode.com/posts"

try:
    response = httpx.get(API_URL)
    response.raise_for_status()  # Raise an exception for bad status codes
    logging.info(f"Successfully fetched data from {API_URL}")
    posts = response.json()
except httpx.RequestError as e:
    logging.error(f"An error occurred while requesting {e.request.url!r}.")
except httpx.HTTPStatusError as e: 
    logging.error(f"Error response {e.response.status_code} while requesting {e.request.url!r}.")
```

### Step 4: Log the results

```python
# Log the title of the first 5 posts
for post in posts[:5]:
    logging.info(f"Post Title: {post['title']}")
```
