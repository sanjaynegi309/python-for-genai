# Module 1.2: Types, Dataclasses, and Pydantic

**Learning Objective:** Understand and use Python's typing system, dataclasses, and Pydantic for creating structured and validated data models.

This notebook introduces type hints, dataclasses for boilerplate-free classes, and Pydantic for data validation.

## Type Hints

Python is a dynamically typed language, but it supports type hints, which can improve code clarity and help catch errors.

```python
def greet(name: str) -> str:
    return f"Hello, {name}"

greet("Alice")
```

## Dataclasses

Dataclasses are a way to create classes that are primarily used for storing data. They automatically generate methods like `__init__`, `__repr__`, and `__eq__`.

```python
from dataclasses import dataclass

@dataclass
class User:
    name: str
    email: str
    is_active: bool = True

user = User(name="Bob", email="bob@example.com")
print(user)
```

## Pydantic

`Pydantic` is a library for data validation and settings management using Python type annotations. It's widely used in web development (especially with FastAPI) and data processing pipelines.

```python
from pydantic import BaseModel

class UserModel(BaseModel):
    name: str
    email: str
    is_active: bool = True

# Pydantic will automatically validate the data
user_data = {"name": "Charlie", "email": "charlie@example.com"}
user_model = UserModel(**user_data)
print(user_model)

# This will raise a validation error
try:
    invalid_user_data = {"name": "David", "email": "not-an-email"}
    UserModel(**invalid_user_data)
except Exception as e:
    print(e)
```

## Recap

- **Type hints** improve code readability and help static analysis tools.
- **Dataclasses** reduce boilerplate code for data-centric classes.
- **Pydantic** provides powerful data validation using type hints.
