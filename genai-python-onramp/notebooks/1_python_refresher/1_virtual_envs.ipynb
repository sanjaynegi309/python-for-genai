# Module 1.1: Virtual Environments & Dependency Management

**Learning Objective:** Understand and use virtual environments to manage project dependencies.

This notebook covers the basics of virtual environments using `venv` and managing dependencies with `pip` and `Poetry`.

## What is a Virtual Environment?

A virtual environment is a self-contained directory tree that contains a Python installation for a particular version of Python, plus a number of additional packages.

**Why use them?**
- **Isolation:** Keep dependencies for different projects separate.
- **Reproducibility:** Ensure that your project can be run by others with the same dependencies.
- **Version Control:** Avoid conflicts between packages that require different versions of the same dependency.

## Using `venv` and `pip`

`venv` is the standard tool for creating virtual environments, and it's included in the Python standard library. `pip` is the standard package installer for Python.

```python
# In your terminal (not in this notebook), you would run the following commands:

# 1. Create a virtual environment named 'venv'
# python -m venv venv

# 2. Activate the virtual environment
# On macOS/Linux:
# source venv/bin/activate
# On Windows:
# .\venv\Scripts\activate

# 3. Install packages
# pip install requests

# 4. Deactivate the virtual environment
# deactivate
```

### `requirements.txt`

To share your dependencies with others, you can create a `requirements.txt` file.

```python
# In your terminal:

# 1. Generate a requirements.txt file
# pip freeze > requirements.txt

# 2. Install dependencies from a requirements.txt file
# pip install -r requirements.txt
```

## Using Poetry

`Poetry` is a modern tool for dependency management and packaging in Python. It uses a `pyproject.toml` file to manage dependencies and provides a more integrated workflow.

```python
# In your terminal:

# 1. Initialize a new project with Poetry
# poetry new my-poetry-project

# 2. Add a dependency
# poetry add requests

# 3. Install dependencies from pyproject.toml
# poetry install

# 4. Activate the virtual environment managed by Poetry
# poetry shell

# 5. Run a script within the virtual environment
# poetry run python my_script.py
```

## Using `uv`

`uv` is an extremely fast Python package installer and resolver, written in Rust. It's designed as a drop-in replacement for `pip` and `pip-tools`.

```python
# In your terminal:

# 1. Install uv
# pip install uv

# 2. Create a virtual environment
# uv venv

# 3. Activate the virtual environment (same as venv)
# source .venv/bin/activate

# 4. Install dependencies
# uv pip install -r requirements.txt
```

## Using `conda`

`conda` is an open-source package and environment management system that runs on Windows, macOS, and Linux. It's very popular in the data science and machine learning communities.

```python
# In your terminal:

# 1. Create a new environment
# conda create --name my-env python=3.9

# 2. Activate the environment
# conda activate my-env

# 3. Install dependencies from a requirements.txt file
# conda install --file requirements.txt

# 4. Deactivate the environment
# conda deactivate
```

## Recap

- **Virtual environments** are essential for isolating project dependencies.
- **`venv`** is the standard, built-in tool for creating virtual environments.
- **`pip`** is the standard package installer.
- **`Poetry`** provides an integrated workflow for dependency management and packaging.
- **`uv`** is a modern, extremely fast alternative to pip.
- **`conda`** is a powerful tool popular in data science for managing environments and packages (including non-Python packages).
