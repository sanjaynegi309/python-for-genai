# Module 1.3: Async Basics with httpx

**Learning Objective:** Understand the basics of asynchronous programming in Python and learn how to make async HTTP requests with `httpx`.

This notebook provides a brief introduction to `asyncio` and demonstrates how to use `httpx` for both synchronous and asynchronous API calls.

## Synchronous vs. Asynchronous

- **Synchronous:** Code executes in a sequence, one line at a time. If a task takes a long time (like a network request), the entire program waits.
- **Asynchronous:** Allows the program to work on other tasks while waiting for a long-running task to complete. This is particularly useful for I/O-bound operations like network requests.

## `async` and `await`

Python's `asyncio` library provides the foundation for asynchronous programming. The `async` and `await` keywords are used to define and work with coroutines.

```python
import asyncio

async def main():
    print("Hello")
    await asyncio.sleep(1)
    print("World")

# In a script, you would run this with asyncio.run(main())
# In a Jupyter notebook, you can just await it
await main()
```

## `httpx`: A modern HTTP client

`httpx` is a modern, fully featured HTTP client for Python. It supports both synchronous and asynchronous requests.

### Synchronous Request with `httpx`

```python
import httpx

response = httpx.get('https://api.github.com')
print(response.status_code)
print(response.headers['content-type'])
```

### Asynchronous Request with `httpx`

```python
import httpx

async def fetch_github_api():
    async with httpx.AsyncClient() as client:
        response = await client.get('https://api.github.com')
        print(response.status_code)
        print(response.headers['content-type'])

await fetch_github_api()
```

## Recap

- **Asynchronous programming** is ideal for I/O-bound tasks.
- **`async` and `await`** are the keywords for working with coroutines.
- **`httpx`** is a versatile library for both sync and async HTTP requests.
