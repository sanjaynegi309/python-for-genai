# Module 1.4: Files, JSON, YAML, and Logging

**Learning Objective:** Learn how to work with files, handle structured data formats like JSON and YAML, manage environment variables, and implement logging.

This notebook covers the essentials of file I/O, data serialization, and application configuration.

## File I/O

Reading from and writing to files is a fundamental task. The `with open(...)` syntax is the recommended way to handle files in Python.

```python
# Write to a file
with open("example.txt", "w") as f:
    f.write("Hello, world!\n")

# Read from a file
with open("example.txt", "r") as f:
    content = f.read()
    print(content)
```

## JSON

JSON (JavaScript Object Notation) is a lightweight data-interchange format. Python's `json` module makes it easy to work with JSON data.

```python
import json

data = {"name": "Alice", "age": 30, "isStudent": False}

# Serialize to a JSON string
json_string = json.dumps(data, indent=4)
print(json_string)

# Deserialize from a JSON string
parsed_data = json.loads(json_string)
print(parsed_data)
```

## YAML

YAML (YAML Ain't Markup Language) is a human-readable data serialization standard. It's often used for configuration files. You'll need to install `PyYAML` to work with YAML in Python (`pip install pyyaml`).

```python
import yaml

yaml_data = """
name: Bob
age: 40
isStudent: true
courses:
  - Math
  - Science
"""

# Deserialize from YAML
parsed_yaml = yaml.safe_load(yaml_data)
print(parsed_yaml)
```

## `.env` files and `python-dotenv`

It's a best practice to store secrets and configuration in environment variables, not in your code. `.env` files provide a convenient way to manage these variables in development. You'll need to install `python-dotenv` (`pip install python-dotenv`).

```python
# Create a .env file with the following content:
# API_KEY=your-secret-api-key

from dotenv import load_dotenv
import os

load_dotenv() # take environment variables from .env.

api_key = os.getenv("API_KEY")
print(f"API Key: {api_key}")
```

## Logging

The `logging` module provides a flexible framework for emitting log messages from your programs.

```python
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

logging.info("This is an info message.")
logging.warning("This is a warning message.")
```

## Recap

- Use `with open(...)` for safe file handling.
- `json` and `yaml` are standard formats for structured data.
- Store secrets in environment variables, not code, using `.env` files in development.
- Use the `logging` module for application logging.
